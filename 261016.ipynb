```python
#1부터 100까지 합을 출력하시오
sum = 0
for i in range(1, 101):
  sum += i
print(sum)
```

```python
#1부터 100까지 짝수의 합을 range()의 증가값을 이용하여 구하시오
sum = 0
for i in range(1, 101):
  if i % 2 == 0:
    sum += i
print(sum)
```

```python
#1부터 100까지 짝수의 합을 range()의 증가값을 이용하여 구하시오 (if 문을 쓰지 말 것)
sum = 0
for i in range(2, 101, 2):
  sum += i
print(sum)
```

```python
#정수 하나를 입력받아 그 수의 구구단을 7 x 1 = 7 형식으로 출력하시오
num = int(input("정수를 입력하세요: "))
for i in range(1, 10):
  print(num, "x", i, "=", num * i)
```

```python
#2단부터 9단까지 구구단을 7 x 1 = 7 형식으로 출력하시오
for i in range(2, 10):
  for j in range(1, 10):
    print(i, "x", j, "=", i * j)
  print("----------")
```

```python
#while문을 이용하여 1부터 100까지 합을 출력하시오
sum = 0
i = 1
while i < 101:
  sum += i
  i += 1
print(sum)
```

```python
#양의 정수를 계속 입력받아 합계를 구하다가 0을 입력하면 반복을 끝내고 합계를 출력하세요. (while True와 break 사용)
total = 0

while True:
  num = int(input("양의 정수를 입력하세요: "))

  if num == 0:
    break
  elif num > 0:
    total += num
  else:
    print("양의 정수가 아닙니다.")

print(total)
```

```python
#1부터 20까지의 수 중 짝수만 담은 리스트를 리스트 내포로 만들어 출력하시오
even = [i for i in range(1, 21) if i % 2 == 0]
print(even)
```

```python
#빈 리스트 cart를 만들고 '우유', '빵'을 추가한 뒤, 맨 앞에 '계란'을 넣고, 끝에 ['사과', '바나나']의 항목을 이어 붙여 출력하시오
cart = []
cart.append('우유')
cart.append('빵')
cart.insert(0, '계란')
cart.extend(['사과', '바나나'])
print(cart)
```

```python
#반복문
scores = [85, 95, 70, 75, 99, 93, 88]
total = 0
max_scores = 0

for i in scores:
  total += i

for i in scores:
  if i > max_scores:
    max_scores = i

print(total)
print(max_scores)
```
