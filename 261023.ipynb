```python
#이번에는 수강생이 사용할 수 있는 언어를 여러 개 입력할 수 있게 합니다.
#한 사람이 여러 언어를 쓸 수 있으므로 언어 이름 여러 개를 콤마(,)로 연결해서 입력합니다. 예: Python,C
#앞의 프로그램은 4가지 언어만 받았지만, 이번에는 언어 수를 제한하지 않고 입력받은 모든 언어의 빈도를 구합니다.

langs = {}

while True:
    s = input('사용 가능한 언어(,로 구분) 또는 종료: ')
    s = s.lower()
    if s == '종료':
        print('종료합니다.')
        break
    for i in s.split(','):
        if i in langs:
            langs[i] += 1
        else:
            langs[i] = 1

for key, value in langs.items():
    print(key, value)
```

```python
#국가와 수도 퀴즈 게임

import random

score = 0

for i in range(5):
    nation = random.choice(list(capitals))
    answer = input(f'{nation}의 수도는?')

    if answer == capitals[nation]:
        print('정답입니다!')
        score += 1
    else:
        print(f'오답입니다. 정답은 {capitals[nation]}입니다.')

print(f'{score}문제 정답.')
```
