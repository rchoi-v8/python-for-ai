```python
#주문 금액 계산기

americano_price = 2000
latte_price = 3000
green_tea_price = 2500

total = 0

americano_num = int(input('아메리카노 주문 수:'))
total += americano_num * americano_price

latte_num = int(input('라떼 주문 수:'))
total += latte_num * latte_price

green_tea_num = int(input('녹차 주문 수'))
total += green_tea_num * green_tea_price

print()
print('---------- 주문 내역 ----------')
print(f'아메리카노 {americano_num}잔 x {americano_price}원 = {americano_num * americano_price:,}원')
print(f'라떼 {latte_num}잔 x {latte_price}원 = {latte_num * latte_price:,}원')
print(f'녹차 {green_tea_num}잔 x {green_tea_price}원 = {green_tea_num * green_tea_price:,}원')
print('-----------------------------')
print(f'합계: {total:,}원')
```
