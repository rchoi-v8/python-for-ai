```python
def asmd(x, y, op):
    if op == '+':
        print(x + y)
    elif op == '-':
        print(x - y)
    elif op == '*':
        print(x * y)
    elif op == '/':
        print(x / y)
    else:
        print('잘못된 연산자입니다.')
        print('+ - * / 입력해 주세요.')

x = int(input('첫 번째 수:'))
y = int(input('두 번째 수:'))
op = input('연산자를 입력하시오:')

asmd(x, y, op)
```

```python
import random

com = random.randint(1, 100)
count = 0

for i in range(10):
    user = int(input('숫자를 입력하세요:'))
    count += 1
               
    if user == com:
        print(f'빙고! 정답은 {com}입니다.')
        print(f'{count}번 만에 맞췄습니다.')
        break
    elif user < com:
        print(f'{user}보다 큽니다.')
    else:
        print(f'{user}보다 작습니다.')
```
